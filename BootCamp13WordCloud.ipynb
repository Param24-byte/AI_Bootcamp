```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = """ Modern geopolitics extends far beyond traditional territorial conflicts and state-to-state military positioning; 
it is deeply intertwined with pressing social crises that directly impact civilian populations worldwide. 
One of the most acute social issues driven by contemporary geopolitics is the escalating global displacement and refugee crisis.
Armed conflicts, proxy wars, and severe political instability in regions like the Middle East, parts of Africa, and Eastern Europe have forced 
tens of millions of people to flee their homes. These displaced populations frequently face perilous journeys, legal limbo, and hostile political 
environments upon reaching neighboring countries or crossing into the West, where immigration has become a fiercely polarized domestic political 
weapon.   Furthermore, resource scarcity and environmental degradation—intensified by both climate change and geopolitical competition—are 
exacerbating global inequality. Access to clean water, arable land, and critical transition minerals required for green technologies are frequently 
weaponized through economic sanctions, trade wars, and territorial posturing. This "resource nationalism" disproportionately harms vulnerable, 
marginalized communities in the Global South, who suffer the worst consequences of environmental collapse despite contributing the least to global 
emissions. Consequently, these compounding pressures fuel widespread food insecurity, inflation, and a widening economic divide between 
hyper-digitized superpowers and developing nations.   Compounding these challenges is the pervasive role of digital authoritarianism and
disinformation campaigns in destabilizing democratic societies. Global powers increasingly utilize cyber operations, social media manipulation,
and information warfare to polarize foreign electorates, undermine public trust in institutions, and suppress human rights activism. As algorithms 
dictate the flow of global news, societal cohesion fractures under the weight of orchestrated polarization. Ultimately, today's geopolitical friction 
is not just a contest of governments, but a deeply human crisis affecting the dignity, safety, and livelihoods of ordinary people globally. """

print (text)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
words = word_tokenize(text.lower())
print(words)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]
# The general syntax for a list comprehension is: [expression for item in iterable if condition]
print(filtered_words)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
word_freq = nltk.FreqDist(words)
```

```python
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)
```

```python
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.show()
```
