### Generate User Login Analysis Dataset

```python
import numpy as np
import pandas as pd
import random
from datetime import datetime, timedelta

# Number of records
num_records = 100

# Generate User IDs
user_ids = [f'user_{i:04d}' for i in range(num_records)]

# Generate Login Times
start_date = datetime(2025, 1, 1, 0, 0, 0)
login_times = [start_date + timedelta(minutes=random.randint(0, 365 * 24 * 60)) for _ in range(num_records)]

# Generate Device Types
device_types = random.choices(['mobile', 'desktop', 'tablet'], k=num_records)

# Generate Login Statuses
login_statuses = random.choices(['success', 'failure'], weights=[0.9, 0.1], k=num_records)

# Generate Locations (simplified)
locations = random.choices(['New York', 'London', 'Tokyo', 'Paris', 'Berlin', 'Sydney'], k=num_records)

# Create a DataFrame
data = {
    'user_id': user_ids,
    'login_time': login_times,
    'device_type': device_types,
    'login_status': login_statuses,
    'location': locations
}
df_login_analysis = pd.DataFrame(data)

# Sort by login time for better time series analysis
df_login_analysis = df_login_analysis.sort_values(by='login_time').reset_index(drop=True)

# Display the entire DataFrame
print("Generated DataFrame:")
display(df_login_analysis)
```

### Save DataFrame to CSV

```python
csv_file_path = 'user_login_analysis.csv'
df_login_analysis.to_csv(csv_file_path, index=False)

print(f"DataFrame successfully saved to {csv_file_path}")
```
