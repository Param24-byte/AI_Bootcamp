````python
# @title Used AI for generation of Data
# Understood the code completely first then wrote it using AI
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Set a random seed for reproducibility
np.random.seed(42)

# --- Define possible values for features ---
num_users = 10
users = [f'user_{i+1:02d}' for i in range(num_users)]

action_types = ['login', 'file_access', 'email_send', 'application_launch', 'data_download', 'system_config_change']
resources = ['server_A', 'server_B', 'confidential_doc.pdf', 'public_report.docx', 'HR_system', 'finance_db', 'CRM_app', 'dev_repo']
locations = ['New York', 'London', 'Tokyo', 'Remote_Home', 'Remote_Cafe']
device_types = ['laptop', 'mobile', 'desktop']
success_statuses = ['success', 'failed']

# --- Generate data points ---
num_records = 1000
data = []

start_time = datetime(2023, 1, 1, 9, 0, 0)

for i in range(num_records):
    # Corrected probabilities to sum to 1
    user = np.random.choice(users, p=[0.2, 0.2, 0.1, 0.1, 0.1, 0.05, 0.05, 0.05, 0.05, 0.10]) # Some users are more active
    timestamp = start_time + timedelta(minutes=np.random.randint(0, 60*24*30)) # Over a month period
    action = np.random.choice(action_types, p=[0.3, 0.25, 0.15, 0.1, 0.1, 0.1])
    resource = np.random.choice(resources)
    ip_address = f'192.168.{np.random.randint(1, 255)}.{np.random.randint(1, 255)}'
    location = np.random.choice(locations)
    device = np.random.choice(device_types)
    success = np.random.choice(success_statuses, p=[0.95, 0.05]) # Mostly successful actions

    # Simulate some 'abnormal' behavior
    is_abnormal = False
    if (action == 'system_config_change' and user == 'user_08') or \
       (action == 'data_download' and 'confidential' in resource and location == 'Remote_Cafe' and np.random.rand() < 0.3) or \
       (success == 'failed' and np.random.rand() < 0.2): # Some failed attempts could be normal, some abnormal
        is_abnormal = True

    data.append({
        'timestamp': timestamp,
        'user_id': user,
        'action_type': action,
        'resource_accessed': resource,
        'ip_address': ip_address,
        'location': location,
        'device_type': device,
        'success_status': success,
        'is_abnormal': is_abnormal
    })

df_cyber_uba = pd.DataFrame(data)

print("Generated Cybersecurity User Behavior Analysis Dataset:")
display(df_cyber_uba.head())
print("\nDataset Info:")
df_cyber_uba.info()
````

````python
import re

# Define a function to extract file extension
def extract_file_extension(resource_name):
    match = re.search(r'\.(?P<extension>[a-zA-Z0-9]+)$', resource_name)
    if match:
        return match.group('extension')
    return 'N/A' # For resources without a clear extension (e.g., servers, systems)

# Check if df_cyber_uba is defined before proceeding
if 'df_cyber_uba' in globals():
    # Apply the function to the 'resource_accessed' column to create a new 'file_extension' column
    df_cyber_uba['file_extension'] = df_cyber_uba['resource_accessed'].apply(extract_file_extension)

    print("DataFrame after extracting file extensions:")
    display(df_cyber_uba[['resource_accessed', 'file_extension']].head())

    print("\nUnique file extensions found:")
    display(df_cyber_uba['file_extension'].value_counts())
else:
    print("Error: DataFrame 'df_cyber_uba' is not defined. Please run the cell that generates it (cell ef42115c) first.")
````
